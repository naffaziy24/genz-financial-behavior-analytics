```python
print("Nilai unik dari kolom 'Est. Monthly Income':")
print(df['Est. Monthly Income'].unique())
```

```python
import re

# Bersihkan karakter aneh dan spasi
df['Est. Monthly Income'] = (
    df['Est. Monthly Income']
    .astype(str)
    .str.replace(r'[^\w\s\.\<\>\-Rp]', '', regex=True)  # hapus karakter aneh (seperti â€“)
    .str.replace(r'\s+', ' ', regex=True)               # rapikan spasi
    .str.strip()                                        # hapus spasi di awal/akhir
)

#  Buat peta nilai kategori standar
income_mapping = {
    '< Rp2.000.000': '< Rp2.000.000',
    'Rp2.000.001 - Rp4.000.000': 'Rp2.000.001 - Rp4.000.000',
    'Rp4.000.001 - Rp6.000.000': 'Rp4.000.001 - Rp6.000.000',
    'Rp6.000.001 - Rp10.000.000': 'Rp6.000.001 - Rp10.000.000',
    'Rp10.000.001 - Rp15.000.000': 'Rp10.000.001 - Rp15.000.000',
    '> Rp15.000.000': '> Rp15.000.000',
    'Unknown': 'Unknown'
}

#  Bersihkan format variasi yang mirip
def normalize_income(val):
    if pd.isna(val) or val.strip() == '':
        return 'Unknown'
    text = val.replace('â€“', '-').replace('–', '-').replace('—', '-').replace('–', '-')
    text = text.replace('Rp.', 'Rp').replace('Rp ', 'Rp').replace('  ', ' ')
    text = text.strip()

    # Cari padanan yang cocok di mapping
    for k in income_mapping:
        if all(part in text for part in re.findall(r'Rp\d', k)):  # cocokkan pola angka
            return income_mapping[k]
    # fallback
    for key in income_mapping.keys():
        if key in text:
            return income_mapping[key]
    return 'Unknown'

df['Est. Monthly Income'] = df['Est. Monthly Income'].apply(normalize_income)
```
